# $M_1$ + $M_2$ + $M_3$ into Transformer+MLP

This scenario considers that $M_i$ could be:

- $M_i \in \{T, V, A\} $, U
- $M_i \in \{T\leftarrow, V\leftarrow, A\leftarrow\} $, UM_N
- $M_i \in \{M_1\leftarrow M_2\} $, CM_
- $M_i \in \{M_1\leftarrow M_2, M_3\}, $ MM_
